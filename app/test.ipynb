```python
import pandas as pd
```

```python
current_recipes = pd.read_csv('data/saved_recipes.csv')
current_recipes
```

```python
recipes_path = 'test.csv'
new_recipe_dict = {
    'Total calories (kcal)': 'portion_calories',
    'Total protein (g)': 'portion_proteins',
    'Cost per portion (£)': 'portion_cost',
    'Cost per 100 grams of protein (£/kg)': 'cost_per_100_g_protein',
    'Calories per gram of protein (kcal/g)': 'kcal_per_100_g_protein',
}
recipe_name='test'
new_recipe_dict['Recipe Name'] = recipe_name
```

```python
new_recipe = pd.DataFrame(new_recipe_dict, index=[0])
new_recipe
```

```python
updated_recipes = pd.concat([current_recipes, new_recipe], axis=0)
updated_recipes
```

```python
updated_recipes.to_csv(recipes_path, index=False)
```
